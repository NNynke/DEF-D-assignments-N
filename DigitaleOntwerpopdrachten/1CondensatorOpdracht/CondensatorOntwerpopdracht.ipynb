#### Vraag 2h
Het elektrische veld een afstand $z$ boven het midden van een uniform geladen schijf met een straal $R$ en oppervlakte lading $\sigma$ wordt gegeven door:
$$ \mathbf{E} = \frac{\sigma}{2\epsilon_0} \bigg[ 1- \frac{z}{\sqrt{R^2 + z^2}}\bigg] \hat{\mathbf{z}}$$
Zie probleem 2.6 in het boek van Griffith.

1. Wat moet de hoogte $z$ zijn boven de schijf zodat het elektrische veld door de schijf 10% is in vergelijking met dat van een oneindig geladen plaat?

##### Antwoord 2h

Als de schijf 100% in vergelijking is met dat van een oneindig geladen plaat, dan geldt er: $$ \mathbf{E} = \frac{\sigma}{2\epsilon_0} \bigg[ 1- 0\bigg] \hat{\mathbf{z}} = \frac{\sigma}{2\epsilon_0}$$

Als de schijf 10% in vergelijking is met dat van een oneindig geladen plaats, dan geldt er: $$ \mathbf{E} = \frac{1}{10} \frac{\sigma}{2\epsilon_0} = \frac{\sigma}{2\epsilon_0} \bigg[ 1- \frac{9}{10}\bigg] \hat{\mathbf{z}} $$
Dus: $$ \frac{z}{\sqrt{R^2 + z^2}} = \frac{9}{10} $$

Dit geeft: $z = \sqrt{\frac{81}{19}}R \approx 2.06R$

Het elektrische veld door een uniform geladen schijf, met straal $R$, is 10% in vergelijking met dat van een oneindig geladen plaat als de hoogte $z$, boven het midden van de schijf, de verdubbelde straal $R$ is van de schijf.

````python
r_axis = r_for_E  # duidelijkere naam
z_axis = z_for_E

idx_r0 = int(np.argmin(np.abs(r_axis - 0.0)))  # dichtst bij r=0

# Binnenzijden
idx_z_inside_top = int(np.argmin(np.abs(z_axis - (z_pos_disk1 - grid_step))))
idx_z_inside_bot = int(np.argmin(np.abs(z_axis - (z_pos_disk2 + grid_step))))

# 2) Ez op de as 
Ez_arr = np.asarray(Ez)

Ez_inside_top = Ez_arr[idx_z_inside_top, idx_r0]
Ez_inside_bot = Ez_arr[idx_z_inside_bot, idx_r0]

# 3) Oppervlaktelading op de binnenzijde
sigma_inside_top = epsilon0 * abs(Ez_inside_top)
sigma_inside_bot = epsilon0 * abs(Ez_inside_bot)

# Neem gemiddelde van beide als verwachte sigma 
sigma_expected = 0.5 * (sigma_inside_top + sigma_inside_bot)

# 4) Totale lading op 1 plaat 
A_disk = np.pi * disk_radius**2
Q_plate = sigma_expected * A_disk

# 5) Totale hoeveelheid lading op beide platen 
Q_both_magnitude_sum = 2 * Q_plate

# 6) Capaciteit 
deltaV = abs(V1 - V2)  # bij +10 en -10 is dit 20 V
C = Q_plate / deltaV

# 7) Printen
print("sigma =", sigma_expected, "C/m^2")
print("Q per plaat =", Q_plate, "C")
print("Q totaal =", Q_both_magnitude_sum, "C")
print("C =", C, "F")
````
