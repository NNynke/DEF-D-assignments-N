## *Opdracht 1*: Sanity check 3

#### Voeg hier een foto van de schets van je plaatcondensator in. Geef daarbij ook duidelijk de waarde van $A$ en $d$ aan.
<img src="schets_condensator_sanitycheck3.png" alt="schets condensator groep 15" width="400"/>

```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
epsilon0 = 8.854e-12 # F/m
epsilon_cardboard = 2.75 # F/m
A = 0.01 # m^2
d_cardboard = 0.001 # m
def Capaciteit(e0,e,a,d):
    C = e0*e*(a/d) 
    return C

# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.
```

```python
# Gebruik je formule om uit te rekenen wat je verwacht dat de capaciteit van jullie vlakke plaat condensatoren
# gaat zijn.

C = Capaciteit(epsilon0, epsilon_cardboard, A, d_cardboard)

print(f"Berekende capaciteit plaatcondensator groep 15 is {C*10**12} pF")

```

## Opdracht 4: Uitgewerkte schets (afmetingen!) en plot van de ontwerpgrafiek.

#### Uitgewerkte  schets van de sensor die we gaan maken:
![uitgewerkte schets](Schets_opdracht4_sensor.png)

```python
# Plot hier de verwachte ontwerpgrafiek van jullie sensoren.

import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

m = np.linspace(0, 1, 100) # kg
epsilon_spons = 6.5
epsilon0 = 8.854e-12 # F/m
A = 0.01 # m^2
d_spons_initial = 3.2*10**-2 # m
g = 9.81 # m/s^2
k = 800 # veerconstante (gokje)

x = m * g / k
d = d_spons_initial - x

Capaciteit_verwacht = Capaciteit(epsilon0, epsilon_spons, A, d)*10**12

plt.figure()
plt.title('Mass VS capacity')
plt.plot(m, Capaciteit_verwacht, 'k', label='Data')
plt.xlabel('$m$ (kg)')
plt.ylabel('$C$ (pF)')
plt.legend()
plt.grid()
plt.show()
```

```python
# Plot hier je oorspronkelijke en je uiteindelijke kalibratiegrafieken, fit daar een bijpassende lijn door
# en zet ook de verwachte ontwerpgrafiek erbij

massa = np.array([0, 100, 439.1, 454.7, 680.6, 764.2, 780.6])*10**-3 # kg
Capacity_Arduino_verbeterd = np.array([71.1, 71.7, 76.1, 75.9, 76.3, 77.9, 78.1]) # pF

plt.figure(figsize=(13,5))
plt.subplot(1,2,1)
plt.title('Mass VS capacity (Arduino data)')
plt.plot(massa, Capacity_Arduino, label='Arduino meting 1') # eerste gemeten waarden van de Arduino
plt.plot(massa, Capacity_Arduino_verbeterd, label='Meting 2 (verbeterd)') # tweede gemeten waarden van de Arduino (na verbeteringen)
plt.xlabel('$m$ (kg)')
plt.ylabel('$C$ (pF)')
plt.legend()
plt.grid()

plt.subplot(1,2,2)
plt.title('Mass VS capacity (verwacht)')
plt.plot(m, Capaciteit_verwacht, label='Verwacht') # de verwachte waarden van de capaciteit bij elke massa (berekend met formule adhv d en A)
plt.xlabel('$m$ (kg)')
plt.ylabel('$C$ (pF)')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()

def linear(x,a,b):
    return a * x + b

popt, pcov = curve_fit(linear, massa, Capacity_Arduino_verbeterd) # curve fitten van data
a_fit, b_fit = popt

plt.figure(figsize=(13,5))
plt.subplot(1,2,1)
plt.title('Mass VS capacity (fit)')
plt.plot(massa, Capacity_Arduino_verbeterd, 'k.', label = 'Data') # de Arduino waarden na de verbetering
plt.plot(massa, linear(massa, *popt), label='Fit') # curve fit van deze data punten
plt.xlabel('$m$ (kg)')
plt.ylabel('$C$ (pF)')
plt.legend()
plt.grid()

# extra data puntjes toevoegen ter verbetering fit (aangezien er eerst een gap was tussen 100 en 450 gram massa)
massa2 = np.array([0, 67.2, 100, 180.5, 280.5, 439.1, 454.7, 680.6, 764.2, 780.6])*10**-3 # kg
Capacity_Arduino_verbeterd2 = np.array([71.1, 71.4, 71.7, 73.8, 73.0, 76.1, 75.9, 76.3, 77.9, 78.1]) # pF

plt.subplot(1,2,2)
plt.title('Mass VS capacity (fit, extra punten)')
plt.plot(massa2, Capacity_Arduino_verbeterd2, 'k.', label = 'Data') # data punten, Arduino, na tweede verbetering
plt.plot(massa2, linear(massa2, *popt), label='Fit') # curve fit van deze data (inclusief extra) punten
plt.xlabel('$m$ (kg)')
plt.ylabel('$C$ (pF)')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

## Opdracht 10: Vul hieronder de conclusie in.

De gemaakte sensor is niet geschikt om de massa van verschillende objecten te meten. Het bereik van de gemeten capaciteit is te klein en de meetpunten volgen de lineaire fitlijn niet mooi. Er zijn verbeteringen uitgevoerd in de iteratiestap (meer meetpunten, zachtere spons en stabielere opstelling). Dit zorgde voor een kleine verbetering, maar niet genoeg om de sensor voor een nuttige toepassing te kunnen gebruiken. Er zit een groot verschil tussen de werkelijkheid en voorspelling. Dit komt waarschijnijk door de ideale aannames van constanten (epsilon van een spons) en de opstelling (perfect gestapelde platen en parallel geplaatste platen) die in de werkelijkheid kunnen verschillen. De vlakke plaat formule was genoeg voor onze specifieke sensor, aangezien als de afstand werd aangepast (verkleint), dan veranderde de capaciteit zoals verwacht (vergroot) en andersom.
